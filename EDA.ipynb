```python
#1. Import libraries and load the data
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('SGJobData.csv')
```

```python
#2. Preview the dataset
display(df.head())
```

```python
#3. Check dataset dimensions
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
#4. Check column names and data types
df.info()
```

```python
#5. Check missing values

missing_summary = pd.DataFrame({
    "missing_count": df.isna().sum(),
    "missing_percentage": (
        df.isna().mean() * 100
    ).round(2)
})

display(missing_summary)
```

```python
#6. Check duplicate values
duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)

display(
    df[df.duplicated(keep=False)]
)
```

```python
#7. Generate descriptive statistics
display(df.describe())
```

```python
#8. Generate descriptive stats in detail

# ==========================================
# 0. Load the Original Data
# ==========================================
# df = pd.read_csv("your_original_file.csv")

print(f"Original Dataset Shape: {df.shape[0]} rows and {df.shape[1]} columns\n")

# ==========================================
# 1. Descriptive Statistics for NUMERICAL Columns
# ==========================================
print("--- Numerical Data Summary ---")
# .describe() defaults to numerical columns (int, float)
# It shows count, mean, standard deviation, min, max, and percentiles (25%, 50%, 75%)
display(df.describe().round(2)) 
print("\n")

# ==========================================
# 2. Descriptive Statistics for TEXT/CATEGORICAL Columns
# ==========================================
print("--- Categorical / Text Data Summary ---")
# include=['object', 'string', 'category'] forces it to look at non-numeric columns
# It shows count, number of unique values, the most frequent value (top), and its frequency (freq)
display(df.describe(include=['object', 'string', 'category']))
print("\n")

# ==========================================
# 3. Quick Look at the "categories" Column Specifically
# ==========================================
column_name = "categories"

print(f"--- Deep Dive: '{column_name}' ---")
# Getting the exact count of missing values and unique variations for your target column
missing_count = df[column_name].isna().sum()
unique_count = df[column_name].nunique(dropna=False) # includes NaN as a unique category

print(f"Total Rows: {len(df)}")
print(f"Missing Values (NaN): {missing_count}")
print(f"Unique Categories (including NaN): {unique_count}")
```
